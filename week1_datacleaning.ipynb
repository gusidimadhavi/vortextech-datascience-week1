```python
#imported required python libraries,pandas is used for loading and cleaning,matplotlib is used for visualizatioin 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#load the dataset into pandas dataframe.this allows dataset to be manipulated and analyzed 
df = pd.read_csv("C:/Users/gusid/Downloads/train.csv")
#display first 5 rows
df.head()
```

```python
#display the dataset information ,including number of columns and rows,datatypes 
df.info()
```

```python
#checking the number  missing valuesin each column to determine which column require cleaning
df.isnull().sum()
```

```python
#remove duplicate records from the data set to improvr data quality
df.drop_duplicates(inplace=True)
```

```python
#verified the datatype of each column to ensure that numerical and categorical column are stored correctly
df.dtypes
```

```python
#created a histogram to visualize the Age Distribution of Titanic passengers.using dropna() before plotting for remove missing values
df['Age'].dropna().hist()
plt.title('Age Distribution of Titanic passengers')
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# created a barchart to compare passengers by class
df['Pclass'].value_counts().plot(kind='bar')
plt.title('Number of passengers by class')
plt.xlabel("passenger class")
plt.ylabel("passenger count")
plt.show()
```

```python
# created a line chart for comparing  average ages across class
avg_age=df.groupby('Pclass')['Age'].mean()
plt.plot(avg_age.index,avg_age.values,marker='o')
plt.title('average age by passenger class')
plt.xlabel("passenger class")
plt.ylabel("Average class")
plt.show()
```

```python
#crfeated a pie chart to show the percentage of distribution of male and female
gender=df['Sex'].value_counts()
plt.pie(gender,labels=gender.index,autopct='%1.1f%%',startangle=90)
plt.title("Gender Distribution of Titanic Passengers")
plt.show()
```
